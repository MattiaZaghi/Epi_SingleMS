```python
import loompy
ds = loompy.connect("/date/gcb/gcb_MZ/multiNanoCT/dev_all.loom")
ds
```

```python
count_matrix_sparse = ds.sparse()
```

```python
import pandas as pd

# Convert the sparse matrix to a dense matrix
count_matrix_dense = count_matrix_sparse.todense()

# Convert the dense matrix to a pandas DataFrame
count_matrix_df = pd.DataFrame(count_matrix_dense)

# Save the DataFrame to a CSV file
count_matrix_df.to_csv("/date/gcb/gcb_MZ/multiNanoCT/embryo_dev_count_matrix.csv", index=False, header=False)
```
